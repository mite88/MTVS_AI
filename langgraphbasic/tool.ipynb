```python
from dotenv import load_dotenv  # 환경 변수 설정 파일 로드 (예: Groq API 키)
from langchain_core.messages import HumanMessage  # 사용자 메시지의 기본 클래스 정의
from langchain_groq import ChatGroq

load_dotenv()  # .env 파일을 통해 세션 및 모델에 필요한 정보 로딩

model = ChatGroq(model="openal/get-oss-20b")  # Groq API 에 연결된 LLM 인스턴스 생성
llm_with_tools = model.bind_tools([get_weater, calculate_tax])  # 외부 함수를 도구로 결합하여 모델에 등록

response = llm_with_tools.invoke([HumanMessage(content="서울 날씨")])
# "서울 날씨" 요청을 수행하고 응답 객체 반환 (날씨/세금 관련 툴이 활성화됨)
print(response)  # 결과 출력
```

```python
TOOL_MAP = {
    "get_weater" :get_weater,
    "calculate_tax" :calculate_tax
}

for tool_call in response.tool_calls:
    selected = TOOL_MAP[tool_call['name']]
    result = selected.invoke(tool_call['args'])
    print(f' 결과:{result}')
```

```python
from langgraph.prebuilt import ToolNode

tool_node = ToolNode([get_weater, calculate_tax])

tool_result = tool_node.invoke({'message':[response]})
print(tool_result)
```
