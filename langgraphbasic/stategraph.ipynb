```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict, List

def node_a(state: TypedDictState):
    print('Node A 시작')
    # 리스트 형태로 데이터를 반환하여 add 리듀서가 동작하게 함
    return {
        'message': ['첫 번째 노드 호출 완료!'],
        'step_count': 1
    }

def node_b(state: TypedDictState):
    print('Node B 시작')
    previous_count = state['step_count']
    print(f'상태(State)로부터 확인한 현재 스탭 카운트 = {previous_count}')

    # 기존 'message' 리스트에 새로운 문자열이 담긴 리스트를 추가함
    return {
        'message': ['두 번째 노드 호출 완료!'],
        'step_count': 1
    }
```

```python
# 상태 갱신 만들기
def append_step_count(curr:int, new:int) -> int :
    return curr +new
```

```python
from typing import Annotated
from operator import add

class TypedDictState(TypedDict):
    message: Annotated[List[str], add]
    step_count : Annotated[int, append_step_count] # int만하면 덮어쒸워짐
```

```python
builder = StateGraph(TypedDictState)

builder.add_node(node_a)
builder.add_node(node_b)

builder.add_edge(START, "node_a")
builder.add_edge("node_a", "node_b")
builder.add_edge("node_a", END)

agent = builder.compile()

result = agent.invoke({
    'message' : [],
    'step_count' :0
})

print(result)
```

- 위에서 그냥 int 만하면 : step_count가 덮여쒸워져서 증가하지 않음 :1
