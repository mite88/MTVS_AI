## ChatModel
- LangChain Runnable을 구현하면 LCEL의 구성요소가 될 수 있음
- ChatModel은 Runnable의 구현체


```python
#uv add langchain-groq
#uv add dotenv
```

```python
import getpass
import os
from dotenv import load_dotenv

load_dotenv() # 환경변수 불러오기

API_KEY = os.getenv("GROQ_API_KEY") or getpass.getpass("GROQ API Key:")
print(API_KEY)
```

```python
from langchain_groq import ChatGroq
# 만약 환경 변수 명이 GROQ_API_KEY라면 apk_key를 굳이 셋팅 없이 객체 생성시점에 자동으로 감지함
model = ChatGroq(
    model="llama-3.1-8b-instant",
                 #groq_api_key=API_KEY
)
model.invoke("안녕하세요 2026년 5월 대한민국 뉴스 탑5")
```
