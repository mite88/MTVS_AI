```python
from langchain_core.runnables import RunnableLambda

append_ten = lambda x:x + 10
print(append_ten(10))

run_append_ten = RunnableLambda(append_ten)
print(type(run_append_ten))

run_double = RunnableLambda(lambda x: x * 2) # 위에를 합쳐서 쓸때...

result1 = run_append_ten.invoke(10) # 실행
print(f'run_append.invoke(10) = {result1}')

result2 = run_append_ten.invoke(5)
result2 = run_double.invoke(result2)
print(f'run_double.invoke(result) = {result2}')
```

## ainvoke()

```python
!uv add nest_asyncio
```

```python
import asyncio
import nest_asyncio
from langchain_core.runnables import RunnableLambda
import time

nest_asyncio.apply()

async def append_ten(x):
    print(f'{x} append_ten 시작')
    await asyncio.sleep(2)
    print(f'{x} append_ten 완료')

    return x + 10

run_async_append = RunnableLambda(append_ten)

async def main():
    start = time.time()

    result = await asyncio.gather(
        run_async_append.ainvoke(1),
        run_async_append.ainvoke(3),
        run_async_append.ainvoke(5),
        run_async_append.ainvoke(7)
    )

    end = time.time()

    print(f'result = {result}')
    print(f'elapsed time = {end - start}')

asyncio.run(main())
```

## batch()

```python
from langchain_core.runnables import RunnableLambda

append_ten = lambda x:x + 10
run_append_ten = RunnableLambda(append_ten)

arr = [_ for _ in range(2, 8, 2)]
print(arr)

result = run_append_ten.batch(arr)
print(result)
```

## stram()

```python
import time
from langchain_core.runnables import RunnableLambda

def print_str(data: str):
    for _ in data:
        print(_, end='', flush=True)
        time.sleep(0.1)
        yield _


print(type(print_str('hello')))

run_print_str = RunnableLambda(print_str)
for _ in run_print_str.stream('hello'):
    pass
```
