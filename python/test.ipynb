```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# 5개 제조사별 생산하는 자동차 종류
#파일 불려오기
mpg = pd.read_csv('data/mpg.csv' , encoding='euc-kr')

manufacturers_boolean =[
    mpg[mpg['manufacturer'] == 'toyota'],
    mpg[mpg['manufacturer'] == 'ford' ],
    mpg[mpg['manufacturer'] == 'audi'],
    mpg[mpg['manufacturer'] == 'hyundai'] ,
    mpg[mpg['manufacturer'] == 'honda']
]    
total = pd.concat(manufacturers_boolean, axis=0)

total
```

```python
# 개수 시각화
# countplot(data=?, x=x축, hue=분할기준(나타내고자 하는것))

# x축 : 제조사, 분할기준 : 자동차종류
sns.countplot(data=total, x='manufacturer', hue='category')
```
