## 정적메서드(클래스, 인스턴스, 스태틱)
- 인스턴스 메서드
    - 일반적인 메서드(인스턴스에 귀속되어있는 메서드...인자 self)

- 클래스 메서드
    -   인스턴스가 아닌 클래스 자체에 귀속되어있는 메서드...데코레이터(`@classmethod`)추가, 인자로 cls 받음

- 스태틱 메서드
    - 클래스, 인스턴스 둘다 귀속되지 않는 메서드...데코레이터(`@staticmethod`) 추가, self, cls 도 안받음
    - 클래스와 논리적으로 관련이 있으나 클래스나 인스턴스의 상태에 접근이 필요없는 기능에 사용하는 메서드

```python
# 인스턴스 메서드
class Rectangle:
    def __init__(self, width, height):
        # 인스턴스 변수 설정: 객체마다 고유한 너비와 높이를 가짐
        self.width = width
        self.height = height

    def area(self):
        # 인스턴스 메서드: 객체의 속성(self)을 사용하여 넓이 계산
        return self.width * self.height

# 객체 생성 (width=2, height=2)
rect = Rectangle(2, 2)
# 메서드 호출 및 결과 출력 (2 * 2 = 4)
print(rect.area())
```

```python
#클래스메서드
class User:
    # 클래스 변수: 모든 User 객체가 공유함
    count: int = 0

    def __init__(self, name):
        self.name = name
        # 생성자가 호출될 때마다 클래스 변수 count를 1 증가시킴
        User.count += 1

    @classmethod
    def get_count(cls):
        # 클래스 메서드: 인스턴스가 아닌 클래스 자체의 상태를 반환함
        return cls.count
```

```python
# static method
class MathHelper:

    @staticmethod
    def add(a, b):
        return a + b

    @staticmethod  # 오타 수정
    def is_even(num):  # self 대신 직접 비교할 값(num)을 인자로 받음
        return num % 2 == 0

# 실행 확인
print(MathHelper.add(1, 2))    # 출력: 3
print(MathHelper.is_even(4))   # 출력: True
```

## 매직메서드(Magic Method)
- 메서드 앞뒤에 이줄 밑줄로 감싸져있는 메서드
    - `__init__`
    - `__del__`
    - `__str__`
    - `__repr__`
    - `__eq__`
    - `__lt__`
    - `__le__`
    - `__gt__`
    - `__ge__`
    - `__hash__`

```python
class Labtop:
    def __init__(self, brand, model, price):
        self.brand = brand
        self.model = model
        self.price = price

    #toString()
    def __str__(self):
        return f"{self.brand} {self.model} {self.price}"

labtop = Labtop('삼성', '갤럭시북프로', 100)
print(labtop)
print(str(labtop))
```

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # 읽기 좋은 표현
    def __str__(self):
        return f"({self.x}, {self.y})"

    # 개발자를 위한 공식적인 문자열을 표현하는 메서드
    def __repr__(self):
        return f"Point(x={self.x}, x={self.y})"

a = Point(1,2)
print(str(a))
print(repr(a))
```

## 연산자 오버라이딩

```python
class Eatable:
    def __init__(self, name):  # name을 받을 수 있도록 수정
        self.name = name

    def __add__(self, other):
        # 섞기 로직
        print(f'{self.name}과 {other.name}을 섞었습니다.')
        return self.name + other.name

class Flour(Eatable):
    pass

class Water(Eatable):
    pass

# 이제 정상적으로 작동합니다.
flour = Flour(name='밀가루')
water = Water(name='물')

a = flour + water  # __add__ 호출
print(a)           # 출력: 밀가루 반죽
```
