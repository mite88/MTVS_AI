## match - case

```python
command = 'exit'

if command == 'exit':
    print('Exiting...')
elif command == 'restart':
    print('Restarting...')

match command:
    case 'exit':
        print('Exiting...')
    case 'restart':
        print('Restarting...')
```

## Iterable(반복 가능한 객체)
- for 루프 등에서 원소를 하나씩 꺼낼 수 있는 객체
- `__iter__()`메서드를 구현한 모든 객체가 이에 해당
- 리스트, 튜플, 문자열, 딕셔너리, 집학 , range, 파일개겣, 제너레이터 같은 객체들

* Iterable랑 Iterator(Iterable에서 iter() 함수로 생성되어 next() 함수로 원소를 하나씩 반환하는 반복자 패턴의 구현 객체)은 다름

```python
my_list = [10,20,30]

my_iter = iter(my_list)
print(my_iter)

while True:
    try:
        print(next(my_iter))
    except StopIteration:
        break
```

```python
class EvenNumber:
    def __init__(self, max_val):
        self.max_val = max_val
        self.current = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.current < self.max_val:
            self.current += 2
            return self.current
        else:
            raise StopIteration

for n in EvenNumber(10):
    print(n, end=' ')
print()
```

## enumerate()
- Iterable 순회 시 원소와 함께 인덱스를 동시에 제공하는 내장 함수

```python
fruits = ['사과', '배', '포도', '딸기']

print(enumerate(fruits))

for idx, fruit in enumerate(fruits):
    print(idx, fruit)
```

## zip
- 두 개 이상의 Iterable을 같은 인덱스의 원소끼리 묶어서 튜플의 형태로 반환하는 내장함수

```python
name = ['PengXin', 'ZhangKun', 'Dali', 'Xiaomi']
age = [20, 21, 22, 23]

for name, age in zip(name, age):
    print(name, age)

print(list(zip(name, age)))

# 길이가짧으면 짧은 길이로 나옴
```
