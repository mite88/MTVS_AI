## 비동기 프로그래밍
- 프로그램이 작업을 수행할 때, 완료될 때까지 마냥 기다리지 않고 다른 작업을 동시에 처리할 수 있게 하는 프로그래밍 방식
- `asyncio` -> 이벤트 루프, 코루틴, 태스크, 퓨처

## 동시성 / 병렬성
- 동시성(Concurrency) : 여러 작업이 동시에 진행되는 것 처럼 보이도록 구성된 실행 모델, 실제로 동시에 실행되는가? x -> 동시에 처리되는것 처럼 관리되는가? -> 잠금, 세마포마, 뮤텍스, 데드락, 기아상태
- 병렬성(Parallelism) : 여러 작업이 물리적으로 동시에 실행되는것, 다중코어 CPU에서 여러개의 프로세서를 사용할때 사용가능한 방식

## asyncio
- Python 표준 라이브러리에서 제공하는 비동기 프로그래밍 프레임워크
- 코루틴(Coroutine)을 실행하고 이벤트 루프를 통해 여러 작업을 스케줄링함

### 구조

#### 코루틴
- `async def` : 정의된 비동기 함수

### 이벤트 루프(Event Loop)

asyncio의 실행 엔진

동기 프로그래밍 -> 함수 호출 -> 호출 스택 -> 위에서 아래로 실행 -> 시간이 오래걸리는 작업 시 -> 기다림 -> 명령을 끝까지 수행

asyncio에서는 코루틴 await -> 이벤트루프에게 제어권 넘김 -> 현재 작업이 다시 실행가능한 상태일때까지 다른작업 수행
