````python
class Flyable:
    def fly(self):
        print("I'm flying")

class Swimmable:
    def swim(self):
        print("I'm swimming")

class Duck(Flyable, Swimmable):
    def do_something(self):
        print("I'm doing something")

duck = Duck()
duck.do_something()

# print() 없이 호출
duck.fly()
duck.swim()
````

````python
class Animal:
    def __init__(self, name):
        self.name = name
        print(f'Animal {self.name} created')

    def introduce(self):
        return f'저는 {self.name} 입니다'

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name) # 부모(Animal)의 __init__ 호출
        self.breed = breed
        print(f'Dog {self.name} created')

    def introduce(self):
        # super()를 사용하여 부모 클래스의 introduce 호출
        base = super().introduce()
        return f'{base} 그리고 저는 {self.breed} 입니다'

class Huskey(Dog):
    def __init__(self, name, breed, color):
        super().__init__(name, breed) # 부모(Dog)의 __init__ 호출
        self.color = color
        print(f'Huskey {self.name} created')

    def introduce(self):
        # super()를 사용하여 부모 클래스(Dog)의 introduce 호출
        base = super().introduce()
        # color 앞에 self.를 붙여야 인스턴스 변수에 접근 가능
        return f'{base} 그리고 털색은 {self.color}입니다'

# 객체 생성 및 실행
h = Huskey('아롱이', '진돗개', '흰색')
print(h.introduce())
````

## MRO(Method Resolution Order)
- 파이썬에서 다중 상속을 사용할 때, 특정 메서드를 호출하면 어떤 클래스부터 탐색할지 결정하는 규칙
- C3 선형화(C3 Linearization) 알고리즘을 사용하여 상속 관계를 일직선으로 정렬하며, 이 순서를 통해 메서드 탐색 우선순위를 정함

    - 자식은 부모보다 먼저 탐색됩니다.
    - 부모가 여러 명일 경우, 클래스 정의 시 나열된 순서(왼쪽에서 오른쪽)대로 탐색됩니다.
    - 부모들의 공통 조상은 가장 마지막에 탐색됩니다. (단, 모든 클래스는 최상위 클래스인 object를 상속받습니다.)

단일 상속 (Single Inheritance)
- 자식 클래스 → 부모 클래스 → 조상 클래스 → object

다중 상속 (Multiple Inheritance)
- C3 선형화(C3 Linearization) 알고리즘를  사용한 일렬(Linear) 순서
- 자식 → 왼쪽 부모 → 오른쪽 부모 → 조상클래스 → object

````python
class A:
    def hello(self):
        print("A")

class B(A):
    def hello(self):
        print("B")

class C(A):
    def hello(self):
        print("C")

class D(B, C):
    pass

print(D().hello())
print(D.__mro__) #mro 순서보기
````

## 추상 클래스

````python
class Shape:
    # 이건 불편해 ㅇㅊㅇ
    def area(self):
        raise NotImplementedError('area method must be implemented by subclass')
````

````python
from abc import ABC, ABCMeta, abstractmethod

# 잘사용안함
#class MyAbstractClass(metaclass=ABCMeta):
#    pass

class Shape(ABC):

    # 상속받는 곳에서 해당 함수 구현안하면 오류남
    @abstractmethod
    def area(self):
        pass

#TypeError: Can't instantiate abstract class Shape without an implementation for abstract method 'area'
#shape = Shape()

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    # 구현함
    def area(self):
        return 3.14 * self.radius * self.radius


print(Circle(5).area())
````

## Mixin
- 독립적인 주체가 되기보다는 다른 클래스에 특정 기능을 추가하기 위해 설계된 작은 클래스 의미(DLC ㅇㅁㅇ)
- 클래스 이름 끝에 Mixin, 다중상속을 활용함

````python
class SoundMixin:
    '''소리내는 기능'''
    def bark(self):
        print(f"{self.name}: Woof!")

class RunMixin:
    '''달리는 기능'''
    def run(self):
        print(f"{self.name}: Running!")

class ValidMixin:
    '''이름 유효성 검사'''
    def is_valid(self, name):
        return len(self.name) > 0

class Dog(SoundMixin, RunMixin, ValidMixin):

    def __init__(self, name, breed):
        self.name = name
        self.breed = breed


dog= Dog("까망이","진돗개")

dog.bark()
dog.run()
````
