## 가변 (위치) 인자
- *변수명, 보통은 *args 로 함

````python
def add_all(*numbers):
    pass

add_all(1,2,3,4,5)

# 튜플로 사용해보기
def append_all(*args):
    print(f'받은 인자 = {args}')
    print(f'인자의 타입 = {type(args)}')

append_all(1,2,3,4,5)

def log(level='info', *msg): # 가급적이면 가변인자는 뒤에쓰는것이 좋음
    print(
        '[{level}] {msg}'
        .format(level=level, msg=' '.join(msg))
    )
````

## 매개변수에 언패킹

````python
def add(a,b,c):
    return a + b + c

val_1 = [1,2,3]
varr_2 = 4,5,6

print(add(*val_1)) # 언패킹
print(add(*varr_2)) # 언패킹
````

````python
def describe(name, age, city):
    print(f"""
이름 : {name}
나이 : {age}
도시 : {city}
""")

info = {'name': 'PengXin', 'age': 21, 'city': 'Seoul'}
describe(**info)
````

````python
def func(a, b, c, d):
    print(a+b+c+d)

positional = [1,2,]
keyword = {'c':3, 'd':4}

func(*positional, **keyword) # 이런것도 가능함
````

````python
def log_wrapper(func, *args, **kwargs):
    print(f'log_wrapper 전처리')
    result = func(*args, **kwargs)
    print(f'log_wrapper 후처리')
    return result


log_wrapper(describe, name='PengXin', age=21, city='Seoul') # **kwargs
log_wrapper(add, 1,2,3) # *args
````
