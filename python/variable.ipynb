```python
list1 = [1, 2, 3, 4, 5]
mixed_list = [1, 'a', True, 3.14]
nested_list = [[1, 2], [3, 4], [5, 6]]
chars_list = list('hello')
fruits_list = ['apple', 'banana', 'orange']
range_list = list(range(1, 6))

print(list1)
print(mixed_list)
print(nested_list)
print(chars_list)

print(fruits_list[1])
print(fruits_list[-2])
print(fruits_list[1:3])
print(fruits_list[::-1]) # 역순

print(range_list)

# 소멸자
del fruits_list[0]
print(fruits_list)
```

## 튜플
- 튜플은 내부의 값을 변경하거나 삭제 할 수 없음

```python
empty_tuple = ()
single_element_tuple = (1)
print(type(single_element_tuple)) # <class 'int'>

single_element_tuple1 = (1,)
print(type(single_element_tuple1)) # 뒤에 컴마가 있으면 튜플이됨 <class 'tuple'>

sth = 10, 20, 30
print(sth)
print(type(sth))

colors= ('red', 'green', 'blue')
# colors[0] = 'yellow' # 'tuple' object does not support item assignment
print(f"colors:{colors}")

# 언패킹
red, green, blue = (1,2,3, )
print(red)
print(green)
print(blue)

#first, middle, last = (1,2,3,4,5) # 갯수가 맞아야함
first, *middle, last = (1,2,3,4,5) # *로하면 시퀀스로묶임
print(first)
print(middle)
print(last)
```

## 집합

```python
number = {1,2,3,4,5}
print(type(number))

num_with_dupli = [1,1,2,2,3,3,3,4,4]
print(list(set(num_with_dupli)))
print(set(num_with_dupli))

A = {1, 2, 3, 4, 5}
B = {5, 6, 7, 8, 9, 10}

print(A | B) # 합집합
print(A & B) # 교집합
print(A - B) # 차집합
print(A ^ B) # 대칭 차집합
```

## 식별 연산자
- is : 메모리값이 같은가?(id())

```python
a = [1,2,3]
b = [1,2,3]
c = a

print(a == b)
print(a is b)
print(a is c)
```

## 바다코끼리 연산자(할당 표현식)
- :=

```python
n = len('파이썬')
if n > 2:
    print(n)

# 위를 아래처럼 할수 있음
if(n := len('파이썬')) > 2:
    print(n)
```
