```python
from langchain_community.document_loaders import CSVLoader  # CSV로더 모듈을 가져옵니다.

# CSV로더를 생성합니다.
spec_column_loader = CSVLoader(
    file_path = "data/Iris.csv",  # 데이터가 저장된 csv파일의 경로입니다.
    encoding = "utf-8",  # 파일이 utf-8 형식으로 읽어올려면 설정해야 합니다.
    source_column="Species",  # 원본 문서의 이름을 지정합니다.
    content_columns=['SepalLengthCm', 'SepalWidthCm']  # 원본 문서에서 내용을 추출할 컬럼입니다.
)

# 데이터 로더를 호출하여 데이터를 불러옵니다.
filtered_docs = spec_column_loader.load();

# 처음 3개의 문서를 출력합니다.
print(filtered_docs[:3])
```

```python
import pandas as pd  # pandas 모듈을 가져옵니다.
from langchain_community.document_loaders import DataFrameLoader  # 데이터 프레임 로더를 가져옵니다.

# Iris.csv 파일에서 데이터를 읽어와서 iris_dataframe에 저장합니다.
iris_dataframe = pd.read_csv('data/Iris.csv', encoding='utf-8')  # CSV 파일을 utf-8로 읽습니다.


# 데이터 프레임 로더를 생성합니다. page_content_column은 'Species'라는 컬럼으로 설정되었습니다.
df_loader = spec_column_loader = DataFrameLoader(
    data_frame= iris_dataframe,  # 읽어온 데이터 프레임을 사용합니다.
    page_content_column='Species'  # 페이지 내용이 저장될 컬럼입니다.
)

# 데이터 로더를 호출하여 데이터를 불러옵니다.
df_docs = df_loader.load()

# 처음 3개의 문서를 출력합니다.
# print(df_docs[:3])

iris_dataframe['combine']=(
    "종"+iris_dataframe['Species'].astype(str)+'\n'+
    "꽃받침 길이"+iris_dataframe['SepalLengthCm'].astype(str)+'\n'+
    "꽃받침 너비"+iris_dataframe['SepalWidthCm'].astype(str)
)

# Loader 설정 및 데이터 로드
df_loader = DataFrameLoader(
    data_frame=iris_dataframe,
    page_content_column='combine'
)

df_docs = df_loader.load()
df_docs[:3]
```
