```python
from langchain_ollama.embeddings import OllamaEmbeddings

embedder = OllamaEmbeddings(model="qwen2.5-coder:7b")
```

```python
import numpy as np

def consine_similarity(a, b):
    return np.dot(a, b) / (
        np.linalg.norm(a) * np.linalg.norm(b)
    )
```

```python
APPLE = "사과 과일"
PEAR = "배 과일"
COMPUTER = "데스크톱 컴퓨터"
MONITOR ="컴퓨터 모니터"
```

```python
doc = [APPLE, PEAR, COMPUTER, MONITOR]

vectors = embedder.embed_documents(doc)
```

```python
apple_pear_result = consine_similarity(vectors[0], vectors[1])
apple_monitor_result = consine_similarity(vectors[0], vectors[3])

print(f'apple_pear_result: {apple_pear_result:.2f} ')
print(f'apple_monitor_result: {apple_monitor_result:.2f} ')
```
