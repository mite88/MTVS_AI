```python
from langchain_ollama.embeddings import OllamaEmbeddings

embedder = OllamaEmbeddings(model="qwen2.5-coder:7b")
```

```python
apple ="사과"

vector = embedder.embed_query(apple)

print(vector)
print(len(vector))
```

```python
docs =['안녕하세요','반갑습니다!']
vectors = embedder.embed_documents(docs)

print(vectors[0][1:5], vectors[1][1:5])
```
